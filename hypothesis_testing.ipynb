# Hypothesis Testing

**Positive reviews**
- H₀: At least 30% of positive reviews are related to the quality of food (p ≥ 0.30).
- Hₐ: Less than 30% of positive reviews are related to the quality of food (p < 0.30).

**Negative reviews**
- H₀: At least 30% of negative reviews are related to customer service (p ≥ 0.30).
- Hₐ: Less than 30% of negative reviews are related to customer service (p < 0.30).

## Load data, positive vs negative

```python
import pandas as pd
import re
import numpy as np

with open('DATA/McDonald_s_Reviews.csv', 'r', encoding='utf-8', errors='replace') as f:
    df = pd.read_csv(f)
df = df.dropna(subset=['review', 'rating'])
```

```python
def parse_rating(r):
    s = str(r).strip().lower()
    match = re.search(r'(\d+)\s*star', s)
    return int(match.group(1)) if match else None

df['rating_num'] = df['rating'].apply(parse_rating)
df = df.dropna(subset=['rating_num'])

# Positive = 4 or 5 stars; Negative = 1 or 2 stars (3 is excluded)
positive = df[df['rating_num'].isin([4, 5])].copy()
negative = df[df['rating_num'].isin([1, 2])].copy()

print('Positive reviews (4–5 stars):', len(positive))
print('Negative reviews (1–2 stars):', len(negative))
```
